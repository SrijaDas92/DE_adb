```python
%sql
drop table if exists chocolate_sales_nw;
create or replace table delta_sharing_catalog.delta_sharing_schema.chocolate_sales_nw as 
select `Sales Person` as Sales_Person, Country, Product, Date, Amount, `Boxes Shipped` as Boxes_Shipped
from delta_sharing_catalog.delta_sharing_schema.chocolate_sales_2
where Country IN ('New Zealand', 'India');

SELECT * FROM delta_sharing_catalog.delta_sharing_schema.chocolate_sales_nw
```

```python
from pyspark.sql import *
from pyspark.sql.types import *
from pyspark.sql.functions import *
```

```python
df = spark.read.table("delta_sharing_catalog.delta_sharing_schema.chocolate_sales_2")
```

```python
df1 = spark.read.table("delta_sharing_catalog.delta_sharing_schema.chocolate_sales_nw")
```

```python
df2 = df.join(df1, df['Sales Person'] == df1['Sales_Person'],"inner")
df3=df1.join(df2, on="Product", how="inner")
#df2.explain()
#df3.explain()
display(df3)
df3.count()
```

broadcastjoin


```python
df_1 = df.join(broadcast(df1),df['Sales Person'] == df1['Sales_Person'], "inner")
#df_1.explain()
display(df_1)
df_1.count()
```

```python
df4=df1.union(df)
df4.display()
df4.count()
df4.printSchema()
```

```python
df4.foreachPartition(lambda x: print(x))
df4.display()
```
