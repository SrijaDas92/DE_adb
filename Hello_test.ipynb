```python
def print_even():
    for x in range(1, 10):
        if x % 2 == 0:
            print(x)
print_even()            
```

```python
def print_odd():
    for x in range(1, 10):
        if x % 2 != 0:
            print(x)
print_odd()
```

```python
def print_prime():
    for n in range(1, 10):
        for x in range(2, n):
            if n % x == 0:
                print(n, 'equals', x, '*', n//x)
                break
        else:
            # loop fell through without finding a factor
            print(n, 'is a prime number')
print_prime()
print_even()
print("Hello World")
```

```python
def print_fibbonachi():
    a, b = 0, 1
    while b < 10:
        print(b)
        a, b = b, a+b
print_fibbonachi()  
```

```python
def print_swap():
    a, b = 0, 1
    a, b = b, a
    print(a, b)
print_swap()
```

```python
def amstrong(n):
    sum = 0
    temp = n
    while temp > 0:
        digit = temp % 10
        sum += digit ** 3
        temp //= 10
    if n == sum:
        return True
    else:
        return False
amstrong(153)
print(amstrong(153))
      
```

```python
def print_factorial():
    def factorial(n):
        if n == 1:
            return 1
        else:
            return n * factorial(n-1)
    print(factorial(5))
print_factorial()
```
