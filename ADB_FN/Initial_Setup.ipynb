```python
%sql
SELECT
    current_catalog() AS current_catalog,
    current_schema() AS current_schema;
```

```python
%sql
CREATE CATALOG IF NOT EXISTS tx_dev
COMMENT 'Development catalog for the Tax Reporting Ingestion Framework';
```

```python
%sql
USE CATALOG tx_dev;
```

```python
%sql
CREATE SCHEMA IF NOT EXISTS tx_dev.bronze
COMMENT 'Raw source data ingested from tax reporting files';

CREATE SCHEMA IF NOT EXISTS tx_dev.control
COMMENT 'Ingestion audit logs, execution status, and control tables';

CREATE SCHEMA IF NOT EXISTS tx_dev.silver
COMMENT 'Reserved for future cleaned and standardized data';

CREATE SCHEMA IF NOT EXISTS tx_dev.gold
COMMENT 'Reserved for future business reporting tables';
```

```python
%sql
SELECT
    current_catalog() AS catalog_name,
    current_schema() AS schema_name;
```
