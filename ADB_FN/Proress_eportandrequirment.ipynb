Build a reusable ingestion framework
Currently, two notebooks contain similar code. In a production-style project, we want to avoid repeating the same ingestion logic for every source.

We'll create a shared Python module that handles:
- Reading CSV and Excel files.
- Standardizing column names.
- Adding ingestion metadata.
- Validating required columns.
- Supporting dry-run mode.
- Writing to Delta tables.
- Recording audit information.
Then each notebook will only need to supply its configuration.

Step 14.1: Add REPLACE_PERIOD support
We'll support three load strategies:
Strategy	Behavior
FULL	Replace the entire target table
APPEND	Add records to the target table
REPLACE_PERIOD	Replace records for one reporting period while preserving other periods


For REPLACE_PERIOD, we'll use tax_year and reporting_period as the partition-selection criteria.
